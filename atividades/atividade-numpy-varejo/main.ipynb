# Atividade Prática: Análise de Dados no Varejo com NumPy


```python
import numpy as np
```

## Nível 1: Básico — Faturamento Semanal e Filtros

### Passo 1: Array de peças vendidas na semana


```python
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
pecas_vendidas
```

### Passo 2: Estimativa de faturamento (operação vetorizada)

A multiplicação por 50 é aplicada a todos os elementos de uma vez, sem laço `for`.


```python
PRECO_MEDIO = 50

faturamento = pecas_vendidas * PRECO_MEDIO
print('Faturamento diário (R$):', faturamento)
print('Faturamento semanal (R$):', faturamento.sum())
```

### Passo 3: Dias de pico (máscara booleana)


```python
mascara_pico = pecas_vendidas > 200
print('Máscara:', mascara_pico)
print('Vendas nos dias de pico:', pecas_vendidas[mascara_pico])
print('Dias de pico (1 = primeiro dia):', np.flatnonzero(mascara_pico) + 1)
```

## Nível 2: Intermediário — Múltiplas Filiais e Falhas de Sistema

### Passo 1: Matriz 2D (lojas nas linhas, dias nas colunas)


```python
lojas = np.array(['Loja A', 'Loja B', 'Loja C'])

vendas_filiais = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330],
])
print(vendas_filiais)
print('Shape:', vendas_filiais.shape)
```

### Passo 2: Total por loja (`axis=1`)

A soma comum propaga o `nan` da Loja A. Para comparação, `np.nansum` ignora a lacuna.


```python
total_por_loja = np.sum(vendas_filiais, axis=1)
total_por_loja_sem_nan = np.nansum(vendas_filiais, axis=1)

for loja, total, total_limpo in zip(lojas, total_por_loja, total_por_loja_sem_nan):
    print(f'{loja}: np.sum = {total} | np.nansum = {total_limpo}')
```

### Passo 3: Média diária geral da rede (ignorando `nan`)


```python
media_contaminada = np.mean(vendas_filiais)
media_geral = np.nanmean(vendas_filiais)

print('np.mean (contaminada):', media_contaminada)
print(f'np.nanmean (correta): {media_geral:.2f} peças por loja/dia')
```

### Passo 4: Meta diária de 200 peças com `np.where()`

Obs.: como `nan >= 200` é `False`, o dia com falha de sistema da Loja A aparece como `'Abaixo'`.


```python
META_DIARIA = 200

status_meta = np.where(vendas_filiais >= META_DIARIA, 'Meta Atingida', 'Abaixo')
print(status_meta)
```

## Nível 3: Avançado — Categorização e Destaques de Marketing

### Passo 1: Amostra de 50 vendas com semente fixa


```python
rng = np.random.default_rng(42)

departamentos = ['Eletrônicos', 'Roupas', 'Casa']
amostra_vendas = rng.choice(departamentos, size=50)
amostra_vendas[:10]
```

### Passo 2: Volume de vendas por departamento (`np.unique`)


```python
categorias, contagens = np.unique(amostra_vendas, return_counts=True)

for categoria, contagem in zip(categorias, contagens):
    print(f'{categoria}: {contagem} vendas')
print('Total:', contagens.sum())
```

### Passo 3: Campanha campeã (`argmax`)


```python
faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])

indice_campea = faturamento_campanhas.argmax()
print(f'Índice da campanha campeã: {indice_campea}')
print(f'Campanha {indice_campea + 1} faturou R$ {faturamento_campanhas[indice_campea]:,}')
```

# Atividade Prática: Organização e Transformação de Dados com NumPy


## Nível 1: Básico — Geração de Sequências e Tipagem de Dados

### Passo 1: Dias do mês com `np.arange`


```python
dias_mes = np.arange(1, 31)
dias_mes
```

### Passo 2: Metas igualmente espaçadas com `np.linspace`


```python
metas_semana = np.linspace(20000, 30000, 5)
metas_semana
```

### Passo 3: Conversão do estoque para inteiros (`.astype(int)`)

O `astype(int)` trunca a parte decimal (não arredonda): `30.9` vira `30`.


```python
estoque_fisico = np.array([10.5, 20.1, 30.9])
estoque_inteiro = estoque_fisico.astype(int)

print(estoque_fisico, estoque_fisico.dtype)
print(estoque_inteiro, estoque_inteiro.dtype)
```

### Passo 4: Últimos 5 dias, do mais recente para o mais antigo


```python
ultimos_5_dias = dias_mes[-5:][::-1]
ultimos_5_dias
```

## Nível 2: Intermediário — Redimensionamento e Proteção de Dados

### Passo 1: 12 meses de visitas em 4 trimestres (`.reshape(4, 3)`)


```python
visitas_mensais = np.array([
    1200, 1350, 1100, 1500, 1620, 1580,
    1700, 1750, 1680, 1900, 2100, 2400,
])

visitas_trimestrais = visitas_mensais.reshape(4, 3)
print(visitas_trimestrais)
print('Shape:', visitas_trimestrais.shape)
```

### Passo 2: Empilhando os semestres com `np.vstack()`

Cada semestre tem 2 trimestres × 3 meses.


```python
semestre_1_marketing = np.array([
    [1200, 1350, 1100],
    [1500, 1620, 1580],
])
semestre_2_comercial = np.array([
    [1700, 1750, 1680],
    [1900, 2100, 2400],
])

visitas_ano = np.vstack((semestre_1_marketing, semestre_2_comercial))
print(visitas_ano)
print('Shape:', visitas_ano.shape)
```

### Passo 3: Projeção do 1º trimestre com `.copy()`

Sem `.copy()`, a fatia é uma _view_: alterar a fatia altera também o array original.


```python
# A "pegadinha": a view compartilha memória com o original
teste_view = visitas_ano.copy()
fatia_view = teste_view[0]
fatia_view[0] = 0
print('Original alterado pela view:', teste_view[0])

# Forma segura: cópia independente
projecao_t1 = visitas_ano[0].copy()
projecao_t1 = (projecao_t1 * 1.10).astype(int)  # cenário de +10%

print('Projeção 1º trimestre:', projecao_t1)
print('Original intacto:     ', visitas_ano[0])
```

### Passo 4: Achatando a matriz com `ravel()`


```python
relatorio_corrido = visitas_ano.ravel()
print(relatorio_corrido)
print('Shape:', relatorio_corrido.shape)
```

## Nível 3: Avançado — Ranking, Broadcasting e Resolução de Sistemas

### Passo 1: Ranking de alunos com `np.argsort()`


```python
notas = np.array([85, 92, 78, 95, 88])

indices_crescente = np.argsort(notas)
indices_ranking = indices_crescente[::-1]  # da maior para a menor nota
print('argsort (crescente):', indices_crescente)
print('Ranking (índices dos alunos):', indices_ranking)

posicao_no_ranking = np.empty_like(indices_ranking)
posicao_no_ranking[indices_ranking] = np.arange(1, len(notas) + 1)

for aluno, (nota, posicao) in enumerate(zip(notas, posicao_no_ranking), start=1):
    print(f'Aluno {aluno}: nota {nota} -> {posicao}º lugar')
```

### Passo 2: Broadcasting de descontos com `np.newaxis`

Preços em coluna `(3, 1)` × descontos em linha `(3,)` resultam numa tabela `(3, 3)`: cada linha é um produto e cada coluna um percentual de desconto.


```python
precos_base = np.array([100, 250, 400])
descontos = np.array([0.05, 0.10, 0.20])

precos_coluna = precos_base[:, np.newaxis]
print('Shape linha:', precos_base.shape, '| Shape coluna:', precos_coluna.shape)

precos_com_desconto = precos_coluna * (1 - descontos)
print(precos_com_desconto)
```

### Passo 3: Custo unitário com `np.linalg.solve()`

- Projeto A: `2·placa + 1·sensor = 500`
- Projeto B: `1·placa − 1·sensor = 100` (sensor devolvido/estornado)


```python
A = np.array([
    [2, 1],
    [1, -1],
])
b = np.array([500, 100])

placa, sensor = np.linalg.solve(A, b)
print(f'Placa robótica: R$ {placa:.2f}')
print(f'Sensor: R$ {sensor:.2f}')
print('Verificação A @ x == b:', np.allclose(A @ np.array([placa, sensor]), b))
```
